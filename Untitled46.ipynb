```python

import pennylane as qml
import numpy as np

# Number of qubits
n_qubits = 4

# Create quantum simulator
dev = qml.device("default.qubit", wires=n_qubits)

# Quantum circuit
@qml.qnode(dev)
def quantum_circuit(inputs, weights):

    # Encode CartPole observations
    qml.AngleEmbedding(
        inputs,
        wires=range(n_qubits),
        rotation="Y"
    )

    # Apply trainable quantum gates
    qml.StronglyEntanglingLayers(
        weights,
        wires=range(n_qubits)
    )

    # Measure quantum outputs
    return [
        qml.expval(qml.PauliZ(i))
        for i in range(n_qubits)
    ]

# Initialize quantum parameters
weights = np.random.uniform(
    0, 2 * np.pi,
    size=(2, n_qubits, 3)
)

# Test with CartPole observation
observation = np.array([
    0.0273956,
    -0.00611216,
    0.03585979,
    0.0197368
])

# Normalize input values
normalized_input = np.array([
    observation[0] / 4.8,
    observation[1] / 3.0,
    observation[2] / 0.418,
    observation[3] / 3.5
]) * np.pi

# Execute circuit
quantum_output = quantum_circuit(
    normalized_input,
    weights
)

print("Number of qubits:", n_qubits)
print("Quantum circuit output:", quantum_output)
print("Quantum circuit created successfully!")
```

```python

import gymnasium as gym
import numpy as np

# Convert CartPole observations into quantum inputs
def preprocess_observation(obs):
    return np.array([
        obs[0] / 4.8,
        obs[1] / 3.0,
        obs[2] / 0.418,
        obs[3] / 3.5
    ]) * np.pi


# Quantum agent selects an action
def quantum_agent_action(obs, weights):

    inputs = preprocess_observation(obs)

    output = np.array(
        quantum_circuit(inputs, weights),
        dtype=float
    )

    # Compare quantum measurement outputs
    action = 1 if output[0] > output[1] else 0

    return action


# Create CartPole environment
env = gym.make("CartPole-v1")

obs, info = env.reset(seed=42)

total_reward = 0

# Test the quantum agent
for step in range(500):

    action = quantum_agent_action(obs, weights)

    obs, reward, terminated, truncated, info = env.step(action)

    total_reward += reward

    if terminated or truncated:
        break

print("Quantum agent test completed!")
print("Total reward:", total_reward)
print("Episode length:", step + 1)

env.close()
```

```python

import time
import numpy as np
import gymnasium as gym

# Evolutionary algorithm settings
POPULATION_SIZE = 10
GENERATIONS = 8
ELITE_SIZE = 2
MUTATION_RATE = 0.15
MUTATION_STRENGTH = 0.25

# Same seeds for fair candidate comparisons
TRAIN_SEEDS = [101, 202]

fitness_evaluations = 0
circuit_evaluations = 0

# Evaluate one quantum candidate
def evaluate_candidate(candidate):

    global fitness_evaluations, circuit_evaluations

    fitness_evaluations += 1
    episode_rewards = []

    for seed in TRAIN_SEEDS:

        env = gym.make("CartPole-v1")
        obs, info = env.reset(seed=seed)

        total_reward = 0

        for step in range(500):

            inputs = preprocess_observation(obs)

            output = np.array(
                quantum_circuit(inputs, candidate),
                dtype=float
            )

            circuit_evaluations += 1

            action = 1 if output[0] > output[1] else 0

            obs, reward, terminated, truncated, info = env.step(action)

            total_reward += reward

            if terminated or truncated:
                break

        episode_rewards.append(total_reward)
        env.close()

    return float(np.mean(episode_rewards))


# Initialize random population
population = [
    np.random.uniform(0, 2 * np.pi, size=(2, 4, 3))
    for _ in range(POPULATION_SIZE)
]

best_weights = None
best_fitness = -np.inf

fitness_history = []

start_time = time.time()

# Evolutionary optimization
for generation in range(GENERATIONS):

    scores = [
        evaluate_candidate(candidate)
        for candidate in population
    ]

    # Sort candidates by fitness
    ranked_indices = np.argsort(scores)[::-1]

    population = [
        population[i] for i in ranked_indices
    ]

    scores = [
        scores[i] for i in ranked_indices
    ]

    # Save best candidate
    if scores[0] > best_fitness:
        best_fitness = scores[0]
        best_weights = population[0].copy()

    fitness_history.append(best_fitness)

    print(
        f"Generation {generation + 1}/{GENERATIONS} | "
        f"Best fitness: {best_fitness:.2f} | "
        f"Generation average: {np.mean(scores):.2f}"
    )

    # Preserve elite candidates
    new_population = [
        population[i].copy()
        for i in range(ELITE_SIZE)
    ]

    # Generate new candidates
    while len(new_population) < POPULATION_SIZE:

        # Select parents from better candidates
        parent1 = population[
            np.random.randint(0, POPULATION_SIZE // 2)
        ]

        parent2 = population[
            np.random.randint(0, POPULATION_SIZE // 2)
        ]

        # Crossover
        child = np.where(
            np.random.rand(2, 4, 3) < 0.5,
            parent1,
            parent2
        )

        # Mutation
        mutation_mask = (
            np.random.rand(2, 4, 3) < MUTATION_RATE
        )

        mutation = np.random.normal(
            0,
            MUTATION_STRENGTH,
            size=(2, 4, 3)
        )

        child = child + mutation_mask * mutation

        # Keep rotation parameters within 0 to 2*pi
        child = np.mod(child, 2 * np.pi)

        new_population.append(child)

    population = new_population


training_time = time.time() - start_time

print("\nEVOLUTIONARY OPTIMIZATION COMPLETED")
print("Best training fitness:", best_fitness)
print("Fitness evaluations:", fitness_evaluations)
print("Circuit evaluations:", circuit_evaluations)
print("Training time:", round(training_time, 2), "seconds")

# Save optimized quantum parameters
np.save("best_quantum_weights.npy", best_weights)
```

```python

import gymnasium as gym
import numpy as np
import time

# Freeze the optimized quantum parameters
frozen_weights = best_weights.copy()

# Evaluation settings
NUM_EVAL_EPISODES = 100
SUCCESS_THRESHOLD = 475

evaluation_returns = []
episode_lengths = []
success_count = 0
evaluation_circuit_calls = 0

inference_times = []

# Evaluate on NEW seeds, not used during training
eval_seeds = range(1000, 1000 + NUM_EVAL_EPISODES)

for seed in eval_seeds:

    env = gym.make("CartPole-v1")
    obs, info = env.reset(seed=seed)

    total_reward = 0
    steps = 0

    while True:

        start = time.perf_counter()

        inputs = preprocess_observation(obs)

        output = np.array(
            quantum_circuit(inputs, frozen_weights),
            dtype=float
        )

        action = 1 if output[0] > output[1] else 0

        inference_times.append(
            time.perf_counter() - start
        )

        evaluation_circuit_calls += 1

        obs, reward, terminated, truncated, info = env.step(action)

        total_reward += reward
        steps += 1

        if terminated or truncated:
            break

    evaluation_returns.append(total_reward)
    episode_lengths.append(steps)

    if total_reward >= SUCCESS_THRESHOLD:
        success_count += 1

    env.close()

# Calculate evaluation metrics
mean_return = np.mean(evaluation_returns)
success_rate = (success_count / NUM_EVAL_EPISODES) * 100
mean_episode_length = np.mean(episode_lengths)

print("========== EVALUATION RESULTS ==========")
print("Evaluation episodes:", NUM_EVAL_EPISODES)
print("Mean evaluation return:", round(mean_return, 2))
print("Success rate:", round(success_rate, 2), "%")
print("Average episode length:", round(mean_episode_length, 2))
print("Successful episodes:", success_count)
print("Quantum circuit evaluations:", evaluation_circuit_calls)
print("Average inference time:", round(np.mean(inference_times), 6), "seconds")
print("========================================")
```

```python

import numpy as np
import gymnasium as gym
import time

# Reproducible evolutionary training
np.random.seed(42)

POPULATION_SIZE = 10
GENERATIONS = 8
ELITE_SIZE = 2
MUTATION_RATE = 0.15
MUTATION_STRENGTH = 0.25

TRAIN_SEEDS = [101, 202]
TARGET_RETURN = 475

environment_steps = 0
steps_to_target = None
fitness_evaluations_target = None

def evaluate_with_step_count(weights, seed):
    global environment_steps

    env = gym.make("CartPole-v1")
    obs, info = env.reset(seed=seed)

    total_reward = 0
    done = False

    while not done:
        action = quantum_agent_action(obs, weights)
        obs, reward, terminated, truncated, info = env.step(action)

        environment_steps += 1
        total_reward += reward
        done = terminated or truncated

    env.close()
    return total_reward

def measured_fitness(weights):
    global fitness_evaluations_target
    fitness_evaluations_target += 1

    scores = [
        evaluate_with_step_count(weights, seed)
        for seed in TRAIN_SEEDS
    ]

    return float(np.mean(scores))

# Create fresh population
population = [
    np.random.uniform(0, 2*np.pi, size=(2, n_qubits, 3))
    for _ in range(POPULATION_SIZE)
]

fitness_evaluations_target = 0
target_found = False

start_time = time.perf_counter()

for generation in range(GENERATIONS):

    scored_population = []

    for weights in population:
        fitness = measured_fitness(weights)
        scored_population.append((fitness, weights.copy()))

        if fitness >= TARGET_RETURN and steps_to_target is None:
            steps_to_target = environment_steps
            fitness_evaluations_target_at_target = fitness_evaluations_target

    scored_population.sort(key=lambda x: x[0], reverse=True)

    best_fitness = scored_population[0][0]

    print(
        f"Generation {generation + 1}/{GENERATIONS} | "
        f"Best fitness: {best_fitness:.2f} | "
        f"Environment steps: {environment_steps}"
    )

    elites = [
        weights.copy()
        for fitness, weights in scored_population[:ELITE_SIZE]
    ]

    new_population = elites.copy()

    while len(new_population) < POPULATION_SIZE:

        parent1 = elites[np.random.randint(len(elites))]
        parent2 = elites[np.random.randint(len(elites))]

        child = np.where(
            np.random.rand(*parent1.shape) < 0.5,
            parent1,
            parent2
        )

        mutation_mask = np.random.rand(*child.shape) < MUTATION_RATE

        child[mutation_mask] += np.random.normal(
            0,
            MUTATION_STRENGTH,
            size=np.sum(mutation_mask)
        )

        child = np.mod(child, 2*np.pi)
        new_population.append(child)

    population = new_population

target_training_time = time.perf_counter() - start_time

print("\n======= ENVIRONMENT STEP RESULTS =======")
print("Target return:", TARGET_RETURN)
print("Environment steps used:", environment_steps)
print("Steps to first reach target:", steps_to_target)
print("Fitness evaluations:", fitness_evaluations_target)
print("Training time:", round(target_training_time, 2), "seconds")
print("========================================")
```
